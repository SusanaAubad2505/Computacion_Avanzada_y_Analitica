```python
import numpy as np
from numpy import linalg
```

## LU

```python
def lu(A):
    
    n=A.shape[0]
    
    U=A.copy().astype(np.double)
    
    L=np.eye(n,dtype=np.double)
    
    for i in range(n-1):
        
        factor = U[i+1:, i] / U[i, i]
        L[i+1:, i] = factor
        U[i+1:] -= factor[:, np.newaxis] * U[i]
        
        
    return L,U
```

## PLU

```python
def plu(A):
    
    n = A.shape[0]
    
    U = A.copy().astype(np.double)
    L = np.eye(n, dtype=np.double)
    P = np.eye(n, dtype=np.double)
    
    
    
#Permutar U para obtener los valores indicados para la diagonal
    for i in range(n-1):
        pivote=np.argmax(np.abs(U[i:,i]))+i
        
        if pivote != i:
            U[[i, pivote]] = U[[pivote, i]]
            P[[i, pivote]] = P[[pivote, i]]
            if i > 0:
                L[[i, pivote], :i] = L[[pivote, i], :i]
            

        factor = U[i+1:, i] / U[i, i]
        L[i+1:, i] = factor
        U[i+1:] -= factor[:, np.newaxis] * U[i]
        
    return P, L, U
```

# Sustitución hacia adelante y hacia atrás

```python
def adelante(L,b):
    
    n=L.shape[0]
    y=np.zeros_like(b,dtype=np.double)
    
    for i in range(n):
        
        y[i]=(b[i]-np.dot(L[i,:i],y[:i]))/L[i,i]
        
    return y


def atras(U,y):
    
    n=U.shape[0]
    x=np.zeros_like(y,dtype=np.double)
    
    for i in range(n - 1, -1, -1):
        
        x[i]=(y[i]-np.dot(U[i,i:],x[i:]))/U[i,i]
        
    return x
```

# Soluciones para LU=A y LU=PA

```python
def solucion_lu(A,b):
    
    L,U=lu(A)
    
    
    y=adelante(L,b)
    
    return atras(U,y)

def solucion_plu(A,b):
    
    P,L,U=plu(A)
    
    Pb=np.dot(P,b)
    
    y=adelante(L,Pb)
    
    return atras(U,y)
    
```
